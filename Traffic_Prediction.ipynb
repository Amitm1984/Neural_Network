```python
import pandas as pd

predictions = pd.DataFrame({
    "cell_id": ["LTE_101", "LTE_102", "LTE_103"],
    "actual_traffic": [72, 55, 81],
    "predicted_traffic": [91, 63, 86]
})

predictions["prediction_time"] = pd.Timestamp.now()
predictions["forecast_minutes"] = 30

predictions["risk"] = predictions[
    "predicted_traffic"
].apply(
    lambda x: "High" if x >= 85
    else "Medium" if x >= 70
    else "Low"
)

print(predictions)
```
