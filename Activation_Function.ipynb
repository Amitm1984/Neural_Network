# Activation Function
 
A neuron first calculated weighted sum of its inputs and then passes that result through Activation Function.

Activation Function introduce nonlinearity, allowing neural network to learn complex pattern such as curved decision boundaries.

### One Important distinction:

An Activation does not necessarily make every operation non linear.

1. Sigmoid Function
2. Tanh Function
3. ReLU Function
4. Softmax Function

## Sigmoid Function
Range:(0,1)
The Sigmoid Function maps with any real-valued input to a number between 0 and 1.

A Large positive input produces value close to 1, and large negative input produces a value close to 0

#### Where it is used?
- Binary Classification Output Layers.
- Multi label Classification, where each label gets its own probability-like score.
- Gates in LSTM and GRUs.

#### Limitations
- Vanishing Gradient. When Sigmoid outputs are close to 0 or 1,its Derivative becomes very small During Back Propagation, gradients can shrink as they travel through many layers, slowing or preventing learning in earlier layers.

## Tanh(Hyperbolic tangent)
Range:(-1,1)
Tanh has an S-shaped curve similar to sigmoid, but its output is centered around zero.

#### Why does Zero- centering matter?
Tanh can reppresent positive, negative and zero valued activations. This can be useful for hidden state representations in recurrent neural networks.

## ReLU(Rectified linear unit)
Ramge:(0,Infinity)
ReLU is one of the most commonly used activation functions in hidden layers.

#### Intution
ReLU supresses negative pre-activations to zero and allows positive pre-activations to pass through unchanged.

Notice That ReLU does not squash every output into fixed range . Its output can be 2,10,100 or any other non negative number.

#### Why ReLU is effective?
- Its Gradient is 1 for positive inputs.
- it avoids sigmoid style saturation on the positive side.
- its Computationaly inexpensive.
- It often work with gradient based optimization in deep networks.

#### Limitation: Dying ReLU

 If a neuron consistentaly recieves negative pre-activations. It ouputs zero and recieves zero Gradient through that Activation.Its weights may stop updating through that path , leaving the neuron inactive.


 ## Softmax 
 
 The Softmax is expecially important for multiclass Classification
 - The Softmax activation function converts a set of raw output scores (called logits, often represented by z) in to probabilities, It is mainly used in the output layers of neural network for multiclass Classification, suppose a neural network predicts wether an image contains cat,dog or horse, its final layer produces three raw scores, called logits.

 Z = [2.0, 1.0,1.0]

 Softmax has two Important Properties:
 - every output is between 0 and 1
 - The outputs sum to 1

```python
from sklearn.datasets import(
    load_breast_cancer,
    load_iris,
    load_diabetes
)

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    mean_squared_error,
    r2_score

)

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
data = load_breast_cancer()
x = data.data
y = data.target
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
scaler =  StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
model = keras.Sequential(
    [layers.Input(shape=(30,)),
     layers.Dense(10,activation='relu'),
     layers.Dense(10,activation='relu'),
     layers.Dense(1,activation='sigmoid')
     ]
)

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
model.summary()
```

```python
model.fit(x_train,y_train,epochs=100)
```

```python
y_pred = model.predict(x_test)
y_pred = (y_pred>0.5).astype(int)
print(accuracy_score(y_test,y_pred))
```

```python
data = load_iris()
x = data.data
y = data.target
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
model = keras.Sequential(
    [layers.Input(shape=(4,)),
     layers.Dense(10,activation='relu'),
     layers.Dense(10,activation='relu'),
     layers.Dense(3,activation='softmax')

    ]
)
model.compile(
    optimizer='adam',
    loss= 'sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
model.summary()
```

```python
model.fit(x_train,y_train,epochs=100)
```

```python
y_pred = model.predict(x_test)
y_pred = y_pred.argmax(axis=1)
print(accuracy_score(y_test,y_pred))
```
