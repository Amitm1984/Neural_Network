# Regularization

A Simple model might learn meaningfull relationships between attendence , study hours and marks. An overfit model might instead assign excessively large weights to small fluctuations or noise in the training data.

## L1 and L2 Regularization

#### L1 regularization(Lasso Like penalty)

Penalize the absolute values of weighhts, It push some weights exactaly to zero,effectively reducing the number of active connection.

#### L2 regularization(Ridge like penalty)

Penalizes squared weights, encouraging them to remain  small rather than  allowing a few weights excessively large


```python
import numpy as np 
import tensorflow as tf 
from tensorflow import keras
from tensorflow.keras import layers,regularizers
```

```python
x =  np.random.randn(1000,10).astype(np.float32)
y = (x[:,0] +0.8 * x[:,1] - 0.5 * x[:,2]>0).astype("float32")
```

```python
model = keras.Sequential([
    layers.Input(shape=(10,)),
    layers.Dense(
        32,
        activation="relu",
        kernel_regularizer=regularizers.l2(0.01)

        ),
        layers.Dense(
            16,
            activation="relu",
            kernel_regularizer=regularizers.l2(0.01)
        ),
        layers.Dense(
            1,
            activation="sigmoid"
        )
])
```

```python
model.summary()
```

```python
x
```

```python
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
model.fit(x,y,epochs=20,verbose=1,batch_size=32)
```
