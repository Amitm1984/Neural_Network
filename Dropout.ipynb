# Dropout:

Dropout is a regularization technique used in neural networks to prevent overfitting.

It randomly turns off some neurons during training so the network does not depend too heavily on particular neurons.

Suppose a hidden layer has 100 neurons.During training , a dropout rate of 0.3 randomely disables approximately 30% of units one each training  step.

This forces to the network to learn useful patterns without depending too heavily on a fixed group of neurons.

```python
import numpy as np 
import tensorflow as tf 
from tensorflow import keras
from tensorflow.keras import layers
```

```python
x = np.random.randn(1200,10).astype("float32")
y = (x[:,0] + x[:,1] - x[:,2] > 0).astype("float32")
```

```python
x_train,x_val = x[:900],x[900:]
y_train,y_val = y[:900],y[900:]
```

```python
model = keras.Sequential([
    layers.Input(shape=(10,)),
    layers.Dense(128,activation='relu'),
    layers.Dropout(0.3),
    layers.Dense(64,activation='relu'),
    layers.Dropout(0.2),
    layers.Dense(1,activation='sigmoid')
])
```

```python
model.summary()
```

```python
model.compile(
    optimizer='adam',
    loss = 'binary_crossentropy',
    metrics = ['accuracy']
)
```

```python
model.fit(x_train,y_train,epochs=100,validation_data=(x_val,y_val))
```
